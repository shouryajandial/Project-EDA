```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import MinMaxScaler, StandardScaler, RobustScaler


height = [120, 122, 125, 127, 121, 123, 137, 131, 161, 145, 141, 132]

#
bins = [118, 125, 135, 160, 200]

labels = ['118 to 125', '126 to 135', '136 to 160', '160 and higher']


category = pd.cut(height, bins=bins, labels=labels, right=True)

print("Original height data:")
print(height)
print("\nCategorized height data:")
print(category)
print("\nValue counts for each category:")
print(category.value_counts().sort_index())
```

```python
print("\n--- Categorizing Height Data using pd.cut with custom bins and labels ---\n")
height = [120, 122, 125, 127, 121, 123, 137, 131, 161, 145, 141, 132]

bins_height = [118, 125, 135, 160, 200]
labels_height = ['Short Height', 'Average Height', 'Good Height', 'Tall Height']

height_categories = pd.cut(height, bins=bins_height, labels=labels_height, right=True)

print("Original height values:")
display(pd.Series(height))
print("\nHeight categories:")
display(height_categories)
print("\nCount of values in each height category:")
display(height_categories.value_counts().sort_index())


print("\n--- Categorizing Random Numbers using pd.cut with an integer for equal bins ---\n")
random_numbers_cut = np.random.rand(40)
num_bins_cut = 5

equal_length_bins = pd.cut(random_numbers_cut, num_bins_cut, precision=2)

print(f"Original random numbers (first 5):\n{random_numbers_cut[:5]}")
print(f"\nCategorized into {num_bins_cut} equal-length bins (first 5):")
display(pd.Series(equal_length_bins).head())
print("\nCounts per bin:")
display(equal_length_bins.value_counts().sort_index())


print("\n--- Categorizing Random Numbers using pd.qcut for quantile-based bins ---\n")
random_numbers_qcut = np.random.rand(2000)
num_quantiles = 4

quantile_bins = pd.qcut(random_numbers_qcut, num_quantiles)

print(f"Original random numbers (first 5):\n{random_numbers_qcut[:5]}")
print(f"\nCategorized into {num_quantiles} quantiles (first 5):")
display(pd.Series(quantile_bins).head())
print("\nCounts per quantile (should be roughly equal):")
display(quantile_bins.value_counts().sort_index())
```

```python
print("Loading sales data...")
df = pd.read_csv('https://raw.githubusercontent.com/salemprakash/EDA/main/Data/sales.csv')
print("Sales data loaded successfully!")
print("\nFirst 10 rows of the original DataFrame:")
display(df.head(10))

print("\nCalculating 'TotalPrice'...")
df['TotalPrice'] = df['UnitPrice'] * df['Quantity']
print("'TotalPrice' column added.")
print("\nDataFrame after adding 'TotalPrice' column:")
display(df.head(10))


print("\nFiltering for TotalPrice > 3,000,000...")
TotalTransaction = df["TotalPrice"]
high_total_transactions = TotalTransaction[np.abs(TotalTransaction) > 3000000]
print("Transactions with TotalPrice > 3,000,000:")
display(high_total_transactions)


print("\nFiltering DataFrame for TotalPrice > 6,741,112...")
extremely_high_value_df = df[np.abs(TotalTransaction) > 6741112]
print("DataFrame rows with TotalPrice > 6,741,112:")
display(extremely_high_value_df)
```

```python
# ------Feature scaling and normalization --------


data = {
    'Feature_A': [10, 15, 20, 25, 30, 100],
    'Feature_B': [100, 150, 200, 250, 300, 1500]
}
df = pd.DataFrame(data)
print("Original Data:")
display(df)

min_max_scaler = MinMaxScaler()
df_min_max = pd.DataFrame(
    min_max_scaler.fit_transform(df),
    columns=df.columns
)
print("Min-Max Scaled Data:")
display(df_min_max)


standard_scaler = StandardScaler()
df_standard = pd.DataFrame(
    standard_scaler.fit_transform(df),
    columns=df.columns
)
print("Standardized Data:")
display(df_standard)
```
