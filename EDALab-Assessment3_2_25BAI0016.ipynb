```python
import numpy as np
import pandas as pd
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer, SimpleImputer, KNNImputer


new_df = pd.DataFrame(
    {
        "Age": [28, np.nan, 35, 40, 25, np.nan, 50],
        "Income": [70000, 85000, np.nan, 90000, 60000, 75000, np.nan],
        "Experience": [3, 10, 7, np.nan, 2, 5, 15],
        "Education": ["Bachelors", "Masters", "PhD", np.nan, "Bachelors", "Masters", "PhD"],
    }
)

print("Original DataFrame with Missing Values:")
print(new_df)
print("\n" + "-"*50 + "\n")


mean_imputer = SimpleImputer(strategy="mean")
new_df["Age_Mean"] = mean_imputer.fit_transform(new_df[["Age"]])


median_imputer = SimpleImputer(strategy="median")
new_df["Income_Median"] = median_imputer.fit_transform(new_df[["Income"]])


mode_imputer = SimpleImputer(strategy="most_frequent")
new_df["Education_Mode"] = mode_imputer.fit_transform(new_df[["Education"]]).ravel()


num_cols_mice = ["Age", "Income", "Experience"]
mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df_mice_imputed = new_df[num_cols_mice].copy()
df_mice_imputed.iloc[:, :] = mice_imputer.fit_transform(df_mice_imputed)
new_df["Age_MICE"] = df_mice_imputed["Age"]
new_df["Income_MICE"] = df_mice_imputed["Income"]
new_df["Experience_MICE"] = df_mice_imputed["Experience"]


num_cols_knn = ["Age", "Income", "Experience"]
knn_imputer = KNNImputer(n_neighbors=2)
df_knn_imputed = new_df[num_cols_knn].copy()
df_knn_imputed.iloc[:, :] = knn_imputer.fit_transform(df_knn_imputed)
new_df["Age_KNN"] = df_knn_imputed["Age"]
new_df["Income_KNN"] = df_knn_imputed["Income"]
new_df["Experience_KNN"] = df_knn_imputed["Experience"]

print("DataFrame after applying various imputation techniques:")
print(new_df)
print("\n" + "-"*50 + "\n")

print("Summary of imputations for Age:")
print(new_df[["Age", "Age_Mean", "Age_MICE", "Age_KNN"]].head(7))

print("\nSummary of imputations for Income:")
print(new_df[["Income", "Income_Median", "Income_MICE", "Income_KNN"]].head(7))

print("\nSummary of imputations for Education:")
print(new_df[["Education", "Education_Mode"]].head(7))
```
