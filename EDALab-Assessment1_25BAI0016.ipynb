```python
import pandas as pd
import numpy as n
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv('/content/drive/MyDrive/chess_games.csv')
```

```python
print("Showing First 5 Lines of Data:")
print(df.head(5))

print("\nDescriptive statistics")
print(df.describe())

resp = df[df['opening_response'].notna()]
print(resp)
```

```python
print("Declined Games:\n")
declined_games = df[df['opening_fullname'].str.contains('Declined', na=False)]
display(declined_games.head())
print(f"Total Declined Games: {len(declined_games)}")

print("\nLesser Used Gambits (Top 10 least frequent gambits with at least 2 occurrences):\n")
gambit_openings = df[df['opening_fullname'].str.contains('Gambit', na=False)]['opening_fullname'].value_counts()
lesser_gambits = gambit_openings[gambit_openings > 1].tail(10)
print(f"Number of unique gambits: {gambit_openings.shape[0]}")
print(lesser_gambits)

print("\nExtremely High Rated Games (both players rating over 2200):\n")
high_rated_games = df[(df['white_rating'] > 2200) & (df['black_rating'] > 2200)]
display(high_rated_games.head())
print(f"Total High Rated Games: {len(high_rated_games)}")

print("\nNon-Null Counts for all columns:\n")
print(df.count())

print("\nResigned Games:\n")
resigned_games = df[df['victory_status'] == 'Resign']
display(resigned_games.head())
print(f"Total Resigned Games: {len(resigned_games)}")
```
