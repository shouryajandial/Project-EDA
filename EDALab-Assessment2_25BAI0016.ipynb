```python
import pandas as pd
import numpy as np

df1SE = pd.DataFrame({
    'StudentID': np.arange(31, 31 + 12*2, 2),
    'ScoreSE': np.random.randint(20, 95, size=12)
})


df2SE = pd.DataFrame({
    'StudentID': np.arange(1, 1 + 15*3, 3),
    'ScoreSE': np.random.randint(25, 100, size=15)
})

df1ML = pd.DataFrame({
    'StudentID': np.arange(10, 10 + 14*2, 2),
    'ScoreML': np.random.randint(30, 90, size=14)
})


df2ML = pd.DataFrame({
    'StudentID': np.arange(5, 5 + 10*4, 4),
    'ScoreML': np.random.randint(40, 99, size=10)
})

print("Generated df1SE head:")
print(df1SE.head())
print("\nGenerated df2SE head:")
print(df2SE.head())
print("\nGenerated df1ML head:")
print(df1ML.head())
print("\nGenerated df2ML head:")
print(df2ML.head())

dfSE_combined = pd.concat([df1SE, df2SE], ignore_index=True)
print("\nCombined SE scores (dfSE_combined) head:")
print(dfSE_combined.head())
print(f"Total SE entries: {len(dfSE_combined)}")


dfML_combined = pd.concat([df1ML, df2ML], ignore_index=True)
print("\nCombined ML scores (dfML_combined) head:")
print(dfML_combined.head())
print(f"Total ML entries: {len(dfML_combined)}")

df_inner_merged = dfSE_combined.merge(dfML_combined, how='inner', on='StudentID')
print("\nStudents common to both courses (Inner Merge Results) head:")
print(df_inner_merged.head())
print(f"Total common students after inner merge: {len(df_inner_merged)}")
```

```python
df_outer_merged = dfSE_combined.merge(dfML_combined, how='outer', on='StudentID')

print("Students in either SE or ML (Outer Merge Results) head:")
print(df_outer_merged.head(10))
print(f"Total students after outer merge: {len(df_outer_merged)}")


df_left_merged = dfSE_combined.merge(dfML_combined, how='left', on='StudentID')

print("All SE students with ML scores (if available) - Left Merge Results head:")
print(df_left_merged.head(10))
print(f"Total students after left merge: {len(df_left_merged)}")
```

```python
high_ml_scores = df_inner_merged[df_inner_merged['ScoreML'] > 80]
zq = df_inner_merged[df_inner_merged['ScoreSE'] > 80]
print("Students with ML Score > 80 (from inner merged data):")
print(high_ml_scores)

df_inner_merged['TotalScore'] = df_inner_merged['ScoreSE'] + df_inner_merged['ScoreML']

print("Inner merged data with 'TotalScore' column:")
print(df_inner_merged.head())
```
